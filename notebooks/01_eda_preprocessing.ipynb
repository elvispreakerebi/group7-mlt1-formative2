# 01 EDA and Preprocessing

This notebook summarizes the data understanding and preprocessing evidence. The reusable work lives in `src/`; the notebook calls the same scripts used for reproducibility.

```python
from pathlib import Path
import sys
import os
import subprocess

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT))

print(f"Python: {sys.executable}")

from src.utils import load_config
from src.data import load_dataset, dataset_summary

config = load_config(ROOT / 'config.yaml')
bundle = load_dataset(config)
dataset_summary(bundle, config)
```

```python
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
venv_python = ROOT / '.venv' / 'bin' / 'python'
script = ROOT / 'scripts' / 'run_eda.py'
runner = venv_python if venv_python.exists() else Path(sys.executable)

env = os.environ.copy()
if venv_python.exists():
    env['MPLCONFIGDIR'] = str(ROOT / 'cache' / 'matplotlib')

completed = subprocess.run(
    [str(runner), str(script)],
    cwd=ROOT,
    check=True,
    text=True,
    capture_output=True,
    env=env,
)
print(completed.stdout, end='')
if completed.stderr:
    print(completed.stderr, end='')
```

Key generated outputs are in `reports/tables/` and `reports/figures/`. These include label frequency, label cardinality, source type distribution, text length, and label co-occurrence.
